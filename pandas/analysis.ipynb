# Pandas Practice

This notebook demonstrates basic DataFrame creation and indexing with pandas.

```python
import pandas as pd

data = {
    "name": ["obed", "solomon", "arbaz", "Aun"],
    "marks": [100, 88, 55, 88],
    "city": ["lahore", "karachi", "pindi", "islamabad"]
}

df = pd.DataFrame(data)
print(df)
print(df.iloc[:, 0])
print(df.loc[0])
```
